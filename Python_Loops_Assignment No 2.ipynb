# **Q7:** `while loop` ka use karke user se tab tak numbers input lein jab tak wo `0` enter na karde. End mein saare numbers ka total dikhayein.

````python
# type your code here
numbers = []
while True:
    num = int(input("Enter number press '0' to exit:  "))
    if num != 0:
        numbers.append(num)
    else:
        print("Thank's")
        break

total = sum(numbers)
print("total = ", total) 
````

# **Q8:** Ek list `numbers = [12, 45, 7, 23, 56, 90, 11]` mein se sabse bada (Maximum) number find karein loop ke zariye.

````python
# type your code here
numbers = [12, 45, 7, 23, 56, 90, 11]
maximum = numbers[0]
for a in numbers:
    if a > maximum:
        maximum = a
print(f"{maximum} is maximum")
````

# **Q12:** User se ek number lein aur check karein ke wo **Prime Number** hai ya nahi (using loops).

````python
# type your code here
num = int(input("Enter any number:  "))
pri_num = True
if num <= -1:
    pri_num = False
else:
    for i in range(2, num):
        if num % i == 0:
            pri_num = False
            break
if pri_num:
    print(f"{num} is Prime Number")
else:
    print(f"{num} is not Prime Number")
````

# **Q19:** **Fibonacci Sequence** ke pehle 10 numbers print karein (0, 1, 1, 2, 3, 5, 8...)

````python
# type your code here
n = 10

a = 0
b = 1

for i in range(n):
    print(a, end=" ")

    c = a + b
    a = b
    b = c
````

# **Q20:** Ek 'Number Guessing Game' banayein jahan program ek number (1-20) fix kare aur user ko 3 chances de guess karne ke liye. `while loop` aur `break` ka use karein.

````python
# type your code here
secret_number = 7
chances = 3

while chances > 0:
    guess = int(input("Guess a number (1-20): "))

    if guess == secret_number:
        print("Correct! You win")
        break
    else:
        chances -= 1
        print("Wrong guess! Chances left:", chances)

if chances == 0:
    print("Game Over! The correct number was", secret_number)
````
