**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Write your code here for Q1
class Student():
    def __init__ (self, name, roll_number):
        self.__name = name
        self.__roll_num = roll_number
    def get_name(self):
        return f"Student Name: {self.__name}"
    def get_roll_num(self):
        return f"Roll Number: {self.__roll_num}"
    def set_name(self, new_name):
        if type(new_name) == str:
            self.__name = new_name
        else:
            return "Name must be str"
    def set_roll_num(self, new_roll_number):
        if type(new_roll_number) == int:
            self.__roll_num = new_roll_number
        else:
            return "Roll number must be int"
```

```python
s1 = Student("Abdullah", 54)
```

```python
s1.get_roll_num()
```

```python
s1.set_roll_num(25)
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount():
    def __init__(self, account_name, account_num, initial_deposit):
        self.acc_name = account_name
        self.acc_num = account_num
        self.__balance = initial_deposit
    def set_deposit(self, amount):
        self.__balance += amount
        return f"You deposit {amount} in {self.acc_name} successfully"
    def set_withdraw(self, amount):
        if self.__balance>amount:
            self.__balance -= amount
        else:
            return "Unsufficient Balance!"
        return f"You withdraw {amount} in {self.acc_name} successfully"
    def get_balance(self):
        return f"{self.acc_name} your current balance is {self.__balance}"
```

```python
a1 = BankAccount('Abdullah', 12345678, 500)
```

```python
a1.set_withdraw(100)
```

```python
a1.get_balance()
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
# Write your code here for Q3
class Employee():
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
    def get_salary(self):
        return f"Salary of {self.name} is {self.__salary}"
    def set_salary(self, new_salary):
        if new_salary<0:
            return "Salary must be in positive"
        else:
            self.__salary = new_salary
```

```python
e1 = Employee("Abdullah", 1000000)
```

```python
e1.set_salary(-2000000)
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# Write your code here for Q4
class Employee():
    def __init__ (self, name, id_no, occupation):
        self.__name = name
        self.__id_no = id_no
        self._occupation = occupation
    def get_name(self):
        return f"Employee Name: {self.__name}"
    def get_id_no(self):
        return f"Employee id: {self.__id_no}"
    def set_name(self, new_name):
        if type(new_name) == str:
            self.__name = new_name
        else:
            return "Name must be str"
    def set_id_no(self, new_id_no):
        if type(new_id_no) == int:
            self.__id_no = new_id_no
        else:
            return "ID must be int"
    def get_occupation(self):
        return f"Employee Occupation: {self._occupation}"
    def set_occupation(self, new_occupation):
        self._occupation = new_occupation
```

```python
e1 = Employee("Abdullah", 23, "Manager")
```

```python
e1.set_occupation("Developer")
```

```python
e1.get_occupation()
```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def get_brand(self):
        return f"The brand of this vehicle is {self.brand}"


class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def get_model(self):
        return f"The model of this car is {self.model}"
```

```python
c1 = Car("Honda", "Civic")
```

```python
c1.get_model()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Student():
    def __init__(self, math, english):
        self.math = math
        self.english = english

    def __add__(self, other):
        math = self.math + other.math
        english = self.english + other.english
        combined = math + english
        return combined

Abdullah = Student(100, 80)
Talha = Student(50, 100)

combined = Abdullah + Talha
combined
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15
class Calculator:
    def __init__(self, a, b=0, c=0):
        self.a = a
        self.b = b
        self.c = c
    def add(self):
        print(f"{self.a + self.b + self.c}")

calc = Calculator(5,7,29)
calc.add()
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
# Write your code here for Q16
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
c1 = Computer()
c1.process()
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17
class Laptop(Computer):
    def process(self):
        print('Laptop is runing')
    def greet(self):
        print('salam')
class Desktop(Computer):
    def process(self):
        print('Desktop is runing')
    def greet(self):
        print('salam')
```

```python
c2 = Laptop()
c2.process()
d1 = Desktop()
d1.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod
class Car(ABC):
    def start_engine(self):
        print('Engine is starting')
    @abstractmethod
    def drive(self):
        pass

class tesla(Car):
    def drive(self):
        print('Tesla is driving')

c1 = tesla()
c1.start_engine()
c1.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass
class CreditCardPayment(Payment):
    def pay(self):
        print('Payment through Credit Card')
class UPIPayment(Payment):
    def pay(self):
        print('Payment through UPI')
```
