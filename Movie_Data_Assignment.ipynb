### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import numpy as np

top5 = data.sort_values(by='IMDB_Rating', ascending=False).head(5)

imdb_scaled = top5['IMDB_Rating'] * 10


x = np.arange(len(top5))
width = 0.35
plt.bar(x - width/2, top5['Meta_score'], width, label='Meta Score')
plt.bar(x + width/2, imdb_scaled, width, label='IMDB Rating (Scaled)')
```
